## Text Cleaning and Preprocessing

```python
text = "I loooove NLP!!! It's sooo coool 😎. #AI #Future2025"
```

1) Convert to lowercase

```python
new_text = text.lower()
print(new_text)
```

2. Removing Punctuation

```python
import string
# string library provides punctuation
punct_chars = string.punctuation
punct_chars = set([punct for punct in punct_chars])
print(punct_chars)
```

```python
# now we remove punctuation from our sentence
for char in punct_chars:
    new_text = new_text.replace(char, '')

print(new_text)
```

3. Removing stopwords

```python
# most frequently used words are stop words
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))
print(f"Stop words are : {stop_words}")
```

```python
# converting new_text into words
tokens = new_text.split(' ')
print(tokens)
```

```python
# now removing stop words
tokens_after_stp_word_removal = [word for word in tokens if word not in stop_words]
print(tokens_after_stp_word_removal)
```

4. Stemming 

    Cuts off word endings (heuristic rule-based). Reduce a word to its root form (not always a real word). Just chops off suffixes/prefixes using rules.

    Example:
    
    playing → play

    studies → studi 

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
# Stemming
stemmer = PorterStemmer()
stems = [stemmer.stem(w) for w in tokens_after_stp_word_removal]
stems
```

5. Lemmatization

Converts words to their base dictionary form (lemma). Ensure the result is a valid word.
Uses morphological analysis + dictionary lookup + POS (part-of-speech).

Example:

playing → play ✅

studies → study ✅

```python
# Lemmatization (using NLTK)
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(w) for w in tokens_after_stp_word_removal]
lemmas
```

## Regex 

Regex is Python library which is mostly used for finding patterns in text.

For example : You have doc which contains some text but lot of useful emails and you want to extract all emails out of that. In that case we can define a general email pattern which then matches all the emails.

```python
# Regex is mostly used for pattern matching in text
import regex as re 

# email matching pattern
pattern = r'[a-z0-9]*@[a-z]*.[a-z]*' 

text = "Hey, My name is Umesh Kumar and my email is umeshkumar94628@gmail.com, Here is my friend sujeet mehta with email sujeetmehta01@gmail.com"

matches = re.findall(pattern, text)
print(matches)
```

```python
# lets find a pattern that matches phone number 
text = "Phone numbers in India +91 5625636373, Phone number in US +61 83749336781"
phone_pattern = r'\+\d{1,3}[-\s]?\d{7,15}'
matches = re.findall(phone_pattern, text)
matches
```

## Tokenization

Breaking documents into sentences

Breaking sentences into words

```python
import nltk 
#nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize 

text = "NLTK is a powerful library for NLP. It offers various tokenization methods."
sent_tokens = sent_tokenize(text)
word_tokens = word_tokenize(text)
print(f"Sentence tokens : {sent_tokens}")
print(f"Word tokens : {word_tokens}")
```
