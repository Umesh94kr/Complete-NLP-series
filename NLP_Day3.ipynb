## Bag of Words implementation

-> For each word in vocabulary it counts the number of times word occured in the particular sentence and use it as numerical representation of that sentence

-> CONS : Completely lost the sequential information as it only keeps count of word regardless of the sequence

```python
paragraph = "Beans. I was trying to explain to somebody as we were flying in, that's corn. That's beans. And they were very impressed at my agricultural knowledge. Please give it up for Amaury once again for that outstanding introduction. I have a bunch of good friends here today, including somebody who I served with, who is one of the finest senators in the country and we're lucky to have him, your Senator, Dick Durbin is here. I also noticed, by the way, former Governor Edgar here, who I haven't seen in a long time and somehow he has not aged and I have. And it's great to see you, Governor. I want to thank President Killeen and everybody at the U of I System for making it possible for me to be here today. And I am deeply honored at the Paul Douglas Award that is being given to me. He is somebody who set the path for so much outstanding public service here in Illinois. Now, I want to start by addressing the elephant in the room. I know people are still wondering why I didn't speak at the commencement."
```

```python
# using sentence tokenizer to break paragraph into sentences
import nltk 
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(paragraph)

# printing sentences
for i, sentence in enumerate(sentences):
    print(i, sentence)
```

```python
# using word tokenizer to break sentences into words 
from nltk.tokenize import word_tokenize 

tokenized_sentences = [word_tokenize(sentence) for sentence in sentences]

# printing tokenized sentences
for i, sentence in enumerate(tokenized_sentences):
    print(i, sentence)
```

```python
# Now we'll remove stop words and punctuation
from nltk.corpus import stopwords 
nltk.download('stopwords')

# Stop word 
stop_words = stopwords.words('english')

# punctuations 
import string 
punct_string = string.punctuation
puncts = [punct for punct in punct_string]

# Words not to keep 
stop_words_and_puncts = stop_words + puncts 
print(stop_words_and_puncts)
```

```python
# keeping only meaningfl words in sentences 
preprocessed_sentences = []
for sentence in tokenized_sentences:
    sent = [word.lower() for word in sentence if word not in stop_words_and_puncts]
    preprocessed_sentences.append(sent)

# printing new sentences 
for sentence in preprocessed_sentences:
    print(sentence)
```

```python
# now we'll make a vocabulary 
word2idx = {}
idx2word = {}

set_of_word = set()

idx = 0
for sentence in preprocessed_sentences:
    for word in sentence:
        if word not in set_of_word:
            word2idx[word] = idx 
            idx2word[idx] = word 
            idx += 1
            set_of_word.add(word)

vocab_size = len(word2idx)
print(f"Size of vocabulary : {vocab_size}")
```

```python
# Now in bag of words we represent a sentence by this vocabulary (index represent words and filling count of how many time that word appeared in sentence)
import torch
vectors = []

for sent in preprocessed_sentences:
    sent_tensor = torch.zeros((vocab_size))
    for word in sent:
        sent_tensor[word2idx[word]] += 1 
    vectors.append(sent_tensor)

for sentence in vectors:
    print(sentence)
```

### Pros of BOW
1) Simplicity - Easy to implement
2) Used for various NLP tasks - text classification, sentiment analysis 

### Cons of BOW
1) Order of word gets lost 
2) Data is sparse computationally inefficient
3) Can't capture meaning of word

## N-Grams

In BOW we consider only one word in vocabulary, but here we can consider n words (consecutive) at one index in vocabulary allowing to keep local context

```python
sentences = ['Machine learning love Umesh Kumar.', 'Umesh Kumar loves Machine learning.']
```

```python
# splitting into bi-gram in which n=2

bigram_sent = sentences[0].split(' ')
bigram_sent = [bigram_sent[i:i+2] for i in range(len(bigram_sent)-1)]
bigram_sent
```

```python
unique_bigrams = set()
bigrams_sentences = []
for sentence in sentences:
    sent = sentence.split(' ')
    bigram_sent = [' '.join(sent[i:i+2]) for i in range(len(sent)-1)]
    bigrams_sentences.append(bigram_sent)
    for bigrams in bigram_sent:
        if bigrams not in unique_bigrams:
            unique_bigrams.add(bigrams)

print(f"Total unique bigrams : {len(unique_bigrams)}")
```

```python
print(unique_bigrams)
```

```python
# printing bigram formatted sentences 
for sent in bigrams_sentences:
    print(sent)
```

```python
# now we'll create vector for these sentences 
word2idx = {word:idx for idx, word in enumerate(unique_bigrams)}
idx2word = {idx:word for idx, word in enumerate(unique_bigrams)}

print(word2idx)
```

```python
vectorized_sent = []

for sent in bigrams_sentences:
    sent_tensor = torch.zeros((len(word2idx)))
    for word in sent:
        sent_tensor[word2idx[word]] += 1
    
    vectorized_sent.append(sent_tensor)

for vector in vectorized_sent:
    print(vector)
```
